```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager

driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()))
driver.implicitly_wait(10)
driver.get('http://localhost:3000')

assert driver.title != "", "Page title is empty"
heading = driver.find_element(By.TAG_NAME, 'h1')
assert heading.is_displayed(), "Header not visible"

login_btn = driver.find_element(By.XPATH, "//button[contains(.,'Sign In') or contains(.,'Login') or contains(.,'OFFICIAL SIGN IN')]")
assert login_btn.is_displayed(), "Login button not visible"

print('[OK] Landing Page Load test passed')
driver.quit()
```
