```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager

driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()))
driver.implicitly_wait(10)
driver.get('http://localhost:3000')

# 1) Login as Citizen
driver.find_element(By.XPATH, "//button[contains(.,'Sign In') or contains(.,'Login') or contains(.,'OFFICIAL SIGN IN')]").click()
time.sleep(1)
driver.find_element(By.XPATH, "//form//input[1]").send_keys('citizen.tanvir@example.com')
driver.find_element(By.XPATH, "//form//input[@type='password']").send_keys('demo1234')
driver.find_element(By.XPATH, "//form//button[@type='submit']").click()
time.sleep(2)

# 2) Open Report Crime wizard
driver.find_element(By.XPATH, "//button[contains(.,'Report Crime')]").click()
time.sleep(1)

# 3) Step 1: Crime category already selected by default -> Continue
driver.find_element(By.XPATH, "//button[contains(.,'Continue')]").click()
time.sleep(1)

# 4) Step 2: Fill Location Name (required) -> Continue
loc_input = driver.find_element(By.XPATH, "//input[contains(@placeholder,'Road') or contains(@placeholder,'Block') or contains(@placeholder,'Banani') or contains(@placeholder,'Landmark')]")
loc_input.clear()
loc_input.send_keys("Gulshan 2 Circle, Dhaka")
driver.find_element(By.XPATH, "//button[contains(.,'Continue')]").click()
time.sleep(1)

# 5) Step 3: Fill Headline and Description (required) -> Continue
headline_input = driver.find_element(By.XPATH, "//input[contains(@placeholder,'snatching') or contains(@placeholder,'Armed') or contains(@placeholder,'incident')]")
headline_input.clear()
headline_input.send_keys("Wallet and Smartphone Theft at Gulshan Market")
desc_field = driver.find_element(By.XPATH, "//textarea[contains(@placeholder,'sequence') or contains(@placeholder,'Describe')]")
desc_field.clear()
desc_field.send_keys("My wallet and smartphone were stolen at Gulshan market around 3pm. The suspect fled towards the north exit.")
driver.find_element(By.XPATH, "//button[contains(.,'Continue')]").click()
time.sleep(1)

# 6) Step 4: Evidence upload (optional) -> Continue
driver.find_element(By.XPATH, "//button[contains(.,'Continue')]").click()
time.sleep(1)

# 7) Step 5: Final Review -> Officially Lodge Crime Report
driver.find_element(By.XPATH, "//button[contains(.,'Lodge') or contains(.,'Officially') or contains(.,'Submit')]").click()
time.sleep(3)

# 8) Assert: GD Docket or tracking ID displayed
body_text = driver.find_element(By.TAG_NAME, 'body').text.lower()
assert any(kw in body_text for kw in ["tracking", "sx-", "submitted", "success", "gd", "filed", "docket", "receipt"]), "full crime report submission failed"
print('[OK] Crime Report Submission test passed')
driver.quit()
```
