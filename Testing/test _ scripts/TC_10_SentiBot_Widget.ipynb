```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager

driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()))
driver.implicitly_wait(10)
driver.get('http://localhost:3000')

# 1) Login as Citizen
driver.find_element(By.XPATH, "//button[contains(.,'Sign In') or contains(.,'Login') or contains(.,'OFFICIAL SIGN IN')]").click()
time.sleep(1)
driver.find_element(By.XPATH, "//form//input[1]").send_keys('citizen.tanvir@example.com')
driver.find_element(By.XPATH, "//form//input[@type='password']").send_keys('demo1234')
driver.find_element(By.XPATH, "//form//button[@type='submit']").click()
time.sleep(2)

# 2) Open Sentinel Prime AI assistant widget
bot_btn = driver.find_element(By.XPATH, "//button[@aria-label='Open Sentinel Prime']")
bot_btn.click()
time.sleep(2)

# 3) Assert: Sentinel Prime widget header, online badge, and welcome content are visible
body_text = driver.find_element(By.TAG_NAME, 'body').text
assert 'SENTINEL PRIME' in body_text, 'Sentinel Prime header missing'
assert 'ONLINE' in body_text, 'Sentinel Prime online badge missing'
assert 'Civic & Legal Assistant' in body_text, 'Sentinel Prime welcome content missing'

print('[OK] SentiBot Widget test passed')
driver.quit()
```
