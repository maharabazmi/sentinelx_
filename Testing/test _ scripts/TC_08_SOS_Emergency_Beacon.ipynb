```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager

driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()))
driver.implicitly_wait(10)
driver.get('http://localhost:3000')

# 1) Login as Citizen
driver.find_element(By.XPATH, "//button[contains(.,'Sign In') or contains(.,'Login') or contains(.,'OFFICIAL SIGN IN')]").click()
time.sleep(1)
driver.find_element(By.XPATH, "//form//input[1]").send_keys('citizen.tanvir@example.com')
driver.find_element(By.XPATH, "//form//input[@type='password']").send_keys('demo1234')
driver.find_element(By.XPATH, "//form//button[@type='submit']").click()
time.sleep(2)

# 2) Open SOS Distress Beacon Center
driver.find_element(By.XPATH, "//div[contains(@class,'short-tabs')]//button[.//span[normalize-space()='SOS']]").click()
time.sleep(1)

# 3) Resolve an old test SOS first, if one is still active
body_text = driver.find_element(By.TAG_NAME, 'body').text
if 'Tracking Beacon ID:' in body_text:
    driver.find_element(By.XPATH, "//button[contains(.,'Stand Down Beacon') or contains(.,'I am Safe')]").click()
    time.sleep(2)

# 4) Trigger emergency SOS beacon
driver.find_element(By.XPATH, "//button[contains(.,'TRIGGER EMERGENCY SOS BEACON')]").click()
time.sleep(1)

# 5) Fill location landmark and confirm the emergency broadcast
location_input = driver.find_element(By.XPATH, "//h3[contains(.,'Confirm Emergency SOS')]/following::input[1]")
location_input.clear()
location_input.send_keys('Gulshan 2 Circle, Dhaka')
driver.find_element(By.XPATH, "//button[contains(.,'Confirm & Broadcast SOS')]").click()
time.sleep(3)

# 6) Assert: A new SOS beacon is active and has a tracking ID
body_text = driver.find_element(By.TAG_NAME, 'body').text.lower()
assert 'tracking beacon id:' in body_text, 'SOS beacon tracking ID was not displayed'
assert any(kw in body_text for kw in ['distress signal transmitted', 'sos sent', 'emergency distress beacon active']), 'SOS emergency beacon was not sent'
print('[OK] SOS Emergency Beacon test passed')
driver.quit()
```
