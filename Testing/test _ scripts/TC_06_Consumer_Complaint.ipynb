```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager

driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()))
driver.implicitly_wait(10)
driver.get('http://localhost:3000')

# 1) Login as Citizen
driver.find_element(By.XPATH, "//button[contains(.,'Sign In') or contains(.,'Login') or contains(.,'OFFICIAL SIGN IN')]").click()
time.sleep(1)
driver.find_element(By.XPATH, "//form//input[1]").send_keys('citizen.tanvir@example.com')
driver.find_element(By.XPATH, "//form//input[@type='password']").send_keys('demo1234')
driver.find_element(By.XPATH, "//form//button[@type='submit']").click()
time.sleep(2)

# 2) Open Consumer Dispute wizard
driver.find_element(By.XPATH, "//button[contains(.,'File Dispute')]").click()
time.sleep(1)

# 3) Step 1: Fill Merchant / Shop Name -> Continue
shop_input = driver.find_element(By.XPATH, "//input[contains(@placeholder,'Al-Madina') or contains(@placeholder,'Swapno') or contains(@placeholder,'Merchant')]")
shop_input.clear()
shop_input.send_keys('Meena Bazar, Uttara')
driver.find_element(By.XPATH, "//button[contains(.,'Continue')]").click()
time.sleep(1)

# 4) Step 2: Fill Product Name (Price Gouging is selected by default) -> Continue
product_input = driver.find_element(By.XPATH, "//input[contains(@placeholder,'Soybean') or contains(@placeholder,'Baby Milk') or contains(@placeholder,'Product')]")
product_input.clear()
product_input.send_keys('Soybean Oil 1L')
driver.find_element(By.XPATH, "//button[contains(.,'Continue')]").click()
time.sleep(1)

# 5) Step 3: Fill MRP, paid price, and description (all required for Price Gouging) -> Continue
mrp_input = driver.find_elements(By.XPATH, "//input[@type='number']")[0]
mrp_input.clear()
mrp_input.send_keys('175')
price_paid_input = driver.find_elements(By.XPATH, "//input[@type='number']")[1]
price_paid_input.clear()
price_paid_input.send_keys('205')
desc_field = driver.find_element(By.XPATH, "//textarea")
desc_field.clear()
desc_field.send_keys('The merchant sold the product above its printed MRP and refused to provide a receipt.')
driver.find_element(By.XPATH, "//button[contains(.,'Continue')]").click()
time.sleep(1)

# 6) Step 4: Evidence upload is optional -> Continue
driver.find_element(By.XPATH, "//button[contains(.,'Continue')]").click()
time.sleep(1)

# 7) Step 5: Final Review -> File Formal DNCRP Complaint
driver.find_element(By.XPATH, "//button[contains(.,'File Formal') or contains(.,'DNCRP') or contains(.,'Lodge')]").click()
time.sleep(3)

# 8) Assert: Consumer complaint receipt and tracking number displayed
body_text = driver.find_element(By.TAG_NAME, 'body').text.lower()
assert 'consumer dispute registered with dncrp' in body_text, 'consumer complaint submission failed'
assert 'tracking number:' in body_text, 'consumer complaint tracking number not displayed'
print('[OK] Consumer Complaint test passed')
driver.quit()
```
