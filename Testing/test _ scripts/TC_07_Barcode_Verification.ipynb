```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service as ChromeService
from webdriver_manager.chrome import ChromeDriverManager

driver = webdriver.Chrome(service=ChromeService(ChromeDriverManager().install()))
driver.implicitly_wait(10)
driver.get('http://localhost:3000')

# 1) Login
driver.find_element(By.XPATH, "//button[contains(.,'Sign In') or contains(.,'Login') or contains(.,'OFFICIAL SIGN IN')]").click()
time.sleep(1)
driver.find_element(By.XPATH, "//form//input[1]").send_keys('citizen.tanvir@example.com')
driver.find_element(By.XPATH, "//form//input[@type='password']").send_keys('demo1234')
driver.find_element(By.XPATH, "//form//button[@type='submit']").click()
time.sleep(2)

# 2) Open Barcode Check tab
driver.find_element(By.XPATH, "//button[contains(.,'Barcode') or contains(.,'BSTI')]").click()
time.sleep(1)

# 3) Enter barcode and verify
input_box = driver.find_element(By.XPATH, "//input")
input_box.clear()
input_box.send_keys("8901030491024")
driver.find_element(By.XPATH, "//button[contains(.,'Verify') or contains(.,'Search') or contains(.,'Check')]").click()
time.sleep(2)

# 4) Assert: Result displayed
body_text = driver.find_element(By.TAG_NAME, 'body').text.upper()
assert "AUTHENTIC" in body_text or "VERIFIED" in body_text or "BSTI" in body_text, "barcode verification failed"
print('[OK] Barcode Verification test passed')
driver.quit()
```
